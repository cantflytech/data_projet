Répartition du nombres d'employer de l'hopital par rapport aux services hospitalier public 

```python
#open effectifs_hopital.csv
import pandas as pd
df_effectifs = pd.read_csv('../data_to_discuss/effectifs_hopital.csv', dtype=str)
print(df_effectifs)
```

```python
#transform column 'chiffres' colum to % de répartition
#exclude total Public row
df_effectifs = df_effectifs[df_effectifs['Secteur Public'] != 'Total Public']
df_effectifs['% de répartition'] = df_effectifs['chiffres'].astype(float) / df_effectifs['chiffres'].astype(float).sum() * 100
#add total column
df_effectifs.loc['Total'] = df_effectifs.sum(numeric_only=True)
```

```python
print(df_effectifs)
```

```python
#mainetannt pour un total de 95 000 personnes quels seraient les effectifs par catégorie
total_personnes = 95000
df_effectifs['effectifs pour 95000'] = (df_effectifs['% de répartition'] / 100) * total_personnes
print(df_effectifs)
```

```python
#retire la colonne chiffres
df_effectifs = df_effectifs.drop(columns=['chiffres'])
#et la colone total
df_effectifs = df_effectifs[df_effectifs.index != 'Total']
```

```python

#arrondi la colonne effectifs pour 95000
df_effectifs['effectifs pour 95000'] = df_effectifs['effectifs pour 95000'].round()

#ajoute la colonne total des effectifs pour 95000
total_effectifs_95000 = df_effectifs['effectifs pour 95000'].sum
print(df_effectifs)
```

```python
#save the file to data folder
df_effectifs.to_csv('../data/effectifs_hopital_analyse.csv', index=False)
```
